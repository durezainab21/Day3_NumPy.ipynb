```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print(numbers)
```

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix)
```

```python
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Data Type:", matrix.dtype)
```

```python
print(numbers[0])
print(numbers[2])
print(numbers[-1])
```

```python
print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
```

```python
a = np.array([2, 4, 6])
b = np.array([1, 3, 5])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
print(a + 10)
print(a * 2)
```

```python
marks = np.array([80, 75, 90, 95, 70])

print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```

```python
student_marks = np.array([65, 78, 92, 88, 74])

print("Marks:", student_marks)
print("Average:", np.mean(student_marks))
print("Highest:", np.max(student_marks))
print("Lowest:", np.min(student_marks))
```

```python
marks = np.array([80, 75, 90, 95, 70])

print("Marks greater than 80:")
print(marks[marks > 80])
```

```python
numbers = np.arange(1, 13)

matrix = numbers.reshape(3, 4)

print(matrix)
```

```python
matrix = np.array([
    [10,20,30],
    [40,50,60]
])

print("Row Sum:", np.sum(matrix, axis=1))
print("Column Sum:", np.sum(matrix, axis=0))
```
